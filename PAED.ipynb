```python
from pathlib import Path

DATASETS = Path("Datasets")

print("Dataset root:", DATASETS.resolve())
print("Folders:", [p.name for p in DATASETS.iterdir() if p.is_dir()])
```

```python
# List the available datasets
for category in ("Graphs", "Tables", "Trees"):
    files = sorted((DATASETS / category).rglob("*.paed"))
    print(f"\n{category} ({len(files)} files):")
    for f in files:
        print(f"  {f.relative_to(DATASETS)}  ({f.stat().st_size:,} bytes)")
```

```python
def preview(path: str | Path, max_lines: int = 10) -> None:
    """Print the first lines of a .paed file."""
    path = Path(path)
    print(f"--- {path} ---")
    with path.open("r", encoding="utf-8", errors="replace") as fh:
        for i, line in enumerate(fh):
            if i >= max_lines:
                break
            print(line.rstrip())

# Example: preview the smallest graph dataset
# preview(DATASETS / "Graphs" / "aCityXXS.paed")
```
